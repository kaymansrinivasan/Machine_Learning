## **Natural Language Toolkit (NLTK)**

**Counting Number of Words / Tokens:**

In this section we will see how to use the computer to count the words in a text in a variety of useful ways. 

Let's begin by finding out the length of text3 (Genesis) from start to finish, in terms of the words and punctuation symbols that appear.

```python
len(text4)
```

**Counting Vocabulary / Unique Types:**

How many distinct words does the book of Genesis contain? The vocabulary of a text is just the set of tokens that it uses, since in a set, all duplicates are collapsed together. 

Let's obtain the vocabulary of text3 in a sorted order:

```python
len(set(text4))
```

```python
len(set(text7))
```

```python
len(sorted(set(text4)))
```

*Note: By wrapping sorted() around the Python expression set(text3)  , we obtain a sorted list of vocabulary items, beginning with various punctuation symbols and continuing with words starting with A. All capitalized words precede lowercase words.* 

How many distinct words in text3?

```python
set(text3)
```

*Note: We discover the size of the vocabulary indirectly, by asking for the number of items in the set, and again we can use len to obtain this number. Although it has 44,764 tokens, this book has only 2,789 distinct words, or "word types".*
